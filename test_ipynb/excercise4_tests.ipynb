```python
from pathlib import Path
import subprocess
import sys

PROJECT_ROOT = Path.cwd()
if not (PROJECT_ROOT / 'directedGraphsUtils.py').exists():
    PROJECT_ROOT = PROJECT_ROOT.parent
sys.path.insert(0, str(PROJECT_ROOT))

import directedGraphsUtils as dgu

GRAPH_FILE = PROJECT_ROOT / 'test_input' / 'directed_graph.txt'
WEIGHTS_FILE = PROJECT_ROOT / 'test_input' / 'directed_weights.txt'
print(f'Projekt: {PROJECT_ROOT}')
print(f'Graf: {GRAPH_FILE}')
```

```python
graph = dgu.load_graph_from_file(GRAPH_FILE, 'adj_list')
weights = dgu.read_weights(WEIGHTS_FILE)

print('Lista sąsiedztwa:')
for node, neighbours in enumerate(graph, start=1):
    print(f'{node}: {neighbours}')
print('Wagi:', weights)
```

```python
components = dgu.kosaraju(graph)
print('Silnie spójne składowe:', components)
```

```python
distances, predecessors = dgu.bellman_ford(graph, weights, start=1)
print('Odległości:', distances)
print('Poprzednicy:', predecessors)
```

```python
distance_matrix = dgu.johnson(graph, weights)
for source, distances in distance_matrix.items():
    print(f'{source}: {distances}')
```
